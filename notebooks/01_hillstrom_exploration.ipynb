**What a response model would conclude:** the Mens E-Mail arm has the highest visit rate (18.28%, against 10.62% with no email), so a response model would rank customers like those in that arm highest and target them first.

**Why that does not tell us who to send the email to:** each bar is an average over everyone in the arm, so it says nothing about which individual customers changed behaviour because of the email, and some of those who visited would have visited anyway (10.62% did with no email at all); the gap between arms estimates the average effect, but sending to the customers most likely to visit is not the same as sending to the customers the email actually persuades.
